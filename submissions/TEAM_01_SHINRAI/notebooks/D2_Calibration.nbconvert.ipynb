# D2: Calibration Primitives

This notebook demonstrates the calibration routines for amplitude, frequency, and DRAG coefficient, and shows their convergence to known true values.


```python
import sys
sys.path.append('../q-autopilot-backend/src')
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(2026)
```

## Amplitude, Frequency, and DRAG Calibration

- **Amplitude** is calibrated using error amplification (repeated $\pi$ pulses).
- **Frequency** is calibrated using Ramsey interferometry.
- **DRAG $\beta$** is calibrated by minimizing leakage into $|2\rangle$.

**Sign Convention:** The DRAG coefficient $\beta_{opt}$ is theoretically $\approx 1/|\alpha|$. We use a sign convention such that $\Omega_y = -\beta \dot{\Omega}_x$. Our empirically found optimal $\beta$ is positive.


```python
from q_autopilot_backend.sim.calibration import simulate_amplitude_calibration, simulate_frequency_calibration, simulate_drag_calibration

amp_conv = simulate_amplitude_calibration(seed=2026)
freq_conv = simulate_frequency_calibration(seed=2026, delta_true=-140.0)
drag_conv = simulate_drag_calibration(seed=2026)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4))

# Amplitude
ax1.errorbar(amp_conv.shots_cum, amp_conv.estimate, yerr=amp_conv.sigma, fmt='-o')
ax1.axhline(amp_conv.truth, color='r', linestyle='--')
ax1.set_xscale('log')
ax1.set_title('Amplitude Calibration')
ax1.set_xlabel('Shots')
ax1.set_ylabel('Amplitude Scale')

# Frequency
ax2.errorbar(freq_conv.shots_cum, freq_conv.estimate, yerr=freq_conv.sigma, fmt='-o')
ax2.axhline(freq_conv.truth, color='r', linestyle='--')
ax2.set_xscale('log')
ax2.set_title('Frequency Calibration')
ax2.set_xlabel('Shots')
ax2.set_ylabel('Detuning (kHz)')

# DRAG
ax3.errorbar(drag_conv.shots_cum, drag_conv.estimate, yerr=drag_conv.sigma, fmt='-o')
ax3.axhline(drag_conv.truth, color='r', linestyle='--')
ax3.set_xscale('log')
ax3.set_title('DRAG Calibration')
ax3.set_xlabel('Shots')
ax3.set_ylabel('Beta (ns)')

plt.tight_layout()
plt.show()
```
